```python
# Definição de schema do df_customer:
schema_df_customer = {
    'customer_id' : 'string', 
    'customer_unique_id' : 'string', 
    'customer_zip_code_prefix' : 'string',
    'customer_city' : 'string', 
    'customer_state' : 'string', 
    'data_ingestao' : 'datetime64[ns]'
}


# Código para reatribuir uma tipagem de dados correta no DataFrame:
    # OBS: no caso do ajuste da data, precisamos atribuir uma regra diferente
    # o coerse é para corrigir caso tenha algum dado de data errado no dataframe
for coluna, tipo_esperado in schema_df_customer.items():
    if "datetime" in tipo_esperado:
        df_customer[coluna] = pd.to_datetime(df_customer[coluna], errors="coerce")
    else:
        df_customer[coluna] = df_customer[coluna].astype(tipo_esperado)
```

```python
# Não temos nenhumo LINHA no DataFrame que se repete:
print(f'Total de linhas duplicadas:', df_customer.duplicated().sum())
# Mas nós precisamos analisar se temos algum ID de customer que está se repetindo (ele é a nossa PK):
print(f'Total de "customer_id" duplicados:', df_customer.duplicated(subset=["customer_id"]).sum())

# Ajustando não conformidades dentro da coluna 'customer_city':
    # Convertendo todos os valores da coluna para minúsculo e retirando espaços em branco:
df_customer["customer_city"] = df_customer["customer_city"].str.strip().str.lower()

    # Conferindo quantidade de valores únicos e printando os valores (como são municipios/cidades, o máximo deve ser 5600 +ou-):
print(f'Total de valores únicos:',df_customer['customer_city'].nunique())
print(f'Lista dos estados registrados: --> ', df_customer['customer_city'].unique())    
```

```python
# Ajustando não conformidades dentro da coluna 'customer_state':
    # Convertendo todos os valores da coluna para maiúsculo e retirando espaços em branco:
df_customer["customer_state"] = df_customer["customer_state"].str.strip().str.upper()

    # Conferindo quantidade de valores únicos e printando os valores (como são estados, o máximo deve ser 27):
print(f'Total de valores únicos:',df_customer['customer_state'].nunique())
print(f'Lista dos estados registrados: --> ', df_customer['customer_state'].unique().tolist())

# Aqui estamos apenas padronizando o prefixo do CEP:
df_customer["customer_zip_code_prefix"] = df_customer["customer_zip_code_prefix"].str.zfill(5)
```

```python
# Salvando arquivo na Silver:
df_customer.to_parquet(caminho_silver + 'customers_silver_dataset.parquet', index=False)
```

## DF_PEDIDOS

```python
# Checagem topo da lista:
df_pedidos.head()
```

```python
# Checagem final tabela:
df_pedidos.tail()
```

```python
# Checagem geral da tabela. Quantidade de linhas, valores nulos, tipo de dados:
df_pedidos.info()
```

```python
# Definição de schema:
schema_df_pedidos = {
    'order_id' : 'string', 
    'customer_id' : 'string', 
    'order_status' : 'string', 
    'order_purchase_timestamp' : 'datetime64[ns]',
    'order_approved_at' : 'datetime64[ns]', 
    'order_delivered_carrier_date' : 'datetime64[ns]',
    'order_delivered_customer_date' : 'datetime64[ns]', 
    'order_estimated_delivery_date' : 'datetime64[ns]',
    'data_ingestao' : 'datetime64[ns]'
}

# Ajuste do tipo de dado atual com o esperado no schema:
for coluna, tipo_esperado in schema_df_pedidos.items():
    if "datetime" in tipo_esperado:
        df_pedidos[coluna] = pd.to_datetime(df_pedidos[coluna], errors="coerce")
    else:
        df_pedidos[coluna] = df_pedidos[coluna].astype(tipo_esperado)

# Conferência linhas duplicadas:
print(f'Quantidade de linhas repetidas:', df_pedidos.duplicated().sum())

# Conferência de dados duplicados na coluna 'order_id':
print(f'Quantidade de dados repetidos na coluna "order_id":', df_pedidos['order_id'].duplicated().sum())
```

```python
# Realizando validações da coluna order_status:
df_pedidos["order_status"] = df_pedidos["order_status"].str.strip().str.lower()

# checando quantidade de elementos presentes na coluna:
print(f'Valores únicos coluna order_status:',df_pedidos["order_status"].nunique())
print(f'Lista dos valores únicos presentes em order_status -->> ',df_pedidos["order_status"].unique().tolist())

# Contabilizando quantidade de ocorrência dos elementos:
df_pedidos["order_status"].value_counts()
```

```python
# Criando colunas auxiliares:
    # Criação da coluna 'data_compra':
df_pedidos['data_compra'] = df_pedidos['order_purchase_timestamp'].dt.date
df_pedidos["data_compra"] = df_pedidos["order_purchase_timestamp"].dt.normalize()

    # Criação da coluna 'tempo_de_entrega':
df_pedidos['tempo_entrega'] = (df_pedidos['order_delivered_customer_date'] - df_pedidos['order_purchase_timestamp']).dt.days

    # Criação da coluna 'atrasou_entrega', para verificar se a entrega foi entregue no prazo ou não:
df_pedidos["atrasou_entrega"] = np.where(
    df_pedidos["order_delivered_customer_date"].notna() &
    df_pedidos["order_estimated_delivery_date"].notna(),
    (df_pedidos["order_delivered_customer_date"] > df_pedidos["order_estimated_delivery_date"]).astype(int),
    np.nan
)
```

```python
# Checando os dados de atraso na entrega:
print(df_pedidos["atrasou_entrega"].value_counts(dropna=False))
print('---'*10)

# Checando o tempo de entrega:
print(df_pedidos["tempo_entrega"].describe())
```

```python
# Quantidade de pedidos que demoraram mais que 3 meses para serem entregues:
print(df_pedidos[df_pedidos['tempo_entrega'] > 90]['tempo_entrega'].sort_values(ascending=False).nunique())

# Verificando casos onde a data de entrega é menor que a data de lançamento pedido (seria um erro):
df_pedidos[
    df_pedidos["order_delivered_customer_date"] < df_pedidos["order_purchase_timestamp"]
]
```

```python
# Salvando arquivo na Silver:
df_pedidos.to_parquet(caminho_silver + 'orders_silver_dataset.parquet', index=False)
```

## DF_GEOLOCALIZACAO

```python
# Conferência topo:
df_geolocalizacao.head()
```

```python
# Conferência base:
df_geolocalizacao.tail()
```

```python
# Conferência geral:
df_geolocalizacao.info()
```

```python
schema_df_geolocalizacao = {
    'geolocation_zip_code_prefix' : 'string',
    'geolocation_lat' : 'float',
    'geolocation_lng' : 'float',
    'geolocation_city' : 'string',
    'geolocation_state' : 'string',
    'data_ingestao' : 'datetime64[ns]'
}

# Ajuste da tipagem de dados do arquivo:
for coluna, tipo_esperado in schema_df_geolocalizacao.items():
    if "datetime" in tipo_esperado:
        df_geolocalizacao[coluna] = pd.to_datetime(df_geolocalizacao[coluna], errors="coerce")
    else:
        df_geolocalizacao[coluna] = df_geolocalizacao[coluna].astype(tipo_esperado)

df_geolocalizacao["geolocation_zip_code_prefix"] = df_geolocalizacao["geolocation_zip_code_prefix"].str.zfill(5)
df_geolocalizacao["geolocation_city"] = df_geolocalizacao["geolocation_city"].str.strip().str.lower()
df_geolocalizacao["geolocation_state"] = df_geolocalizacao["geolocation_state"].str.strip().str.upper()
```

```python
df_geolocalizacao.isnull().sum()
```

```python
df_geolocalizacao_silver = (
    df_geolocalizacao.groupby("geolocation_zip_code_prefix", as_index=False)
    .agg({
        "geolocation_lat": "mean",
        "geolocation_lng": "mean",
        "geolocation_city": "first",
        "geolocation_state": "first",
        "data_ingestao": "first"
    })
)
```

```python
# Conferindo dados duplicados de prefixo:
df_geolocalizacao_silver.duplicated(subset=["geolocation_zip_code_prefix"]).sum()
```

```python
print(df_geolocalizacao.shape)
print(df_geolocalizacao_silver.shape)
```

```python
# Salvando arquivo na Silver:
df_geolocalizacao_silver.to_parquet(caminho_silver + 'geolocalization_silver_dataset.parquet', index=False)
```
