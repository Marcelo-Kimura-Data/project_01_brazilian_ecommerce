```python
import pandas as pd
import os
```

```python
# Estou puxando o nome da pasta onde está armazenado os arquivos do projeto:
caminho_raw = r"C:\Users\caldo\OneDrive\Desktop\Projetos_GIT\project_01_brazilian_ecommerce\data\raw"

# Aqui faço uma simples correção das barras, troco a barra '\' pela '/':
caminho = caminho_raw.replace('\\','/') + '/'
print('Caminho base para os arquivos:')
print(caminho)
print()

# Aqui eu listo os arquivos CSV que estão presentes dentro da pasta 'raw':
lista = os.listdir(caminho)
print(lista)
```

```python
# Agora eu irei criar o caminho dos arquivos referente a cada base de dados.
# Basicamente irei fazer um concatenação entre o caminho + nome do arquivo na raw:
caminhos = []
for arquivo in lista:
    caminho_completo = caminho + arquivo
    caminhos.append(caminho_completo)

for arquivo in caminhos:
    print(arquivo)
```

```python
# Agora irei criar os dataframes referente aos seus respectivos arquivos na base:
df_customer = pd.read_csv(caminhos[0])
df_geolocalizacao = pd.read_csv(caminhos[1])
df_pedidos = pd.read_csv(caminhos[2])
df_itens_pedidos = pd.read_csv(caminhos[3])
df_ordem_pagamento = pd.read_csv(caminhos[4])
df_revisao_pedidos = pd.read_csv(caminhos[5])
df_produtos = pd.read_csv(caminhos[6])
df_vendedores = pd.read_csv(caminhos[7])
df_categoria_translacao = pd.read_csv(caminhos[8])
```
